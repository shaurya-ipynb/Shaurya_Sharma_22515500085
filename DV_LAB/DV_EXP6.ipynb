```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('DV_EXP6.csv')
```

```python
medals = df.dropna(subset=['Medal'])
counts = medals['Team'].value_counts()

top7 = counts.head(7)
pie_data = pd.concat([top7, pd.Series({'Other': counts.iloc[7:].sum()})])

top5 = counts.head(5)
donut_data = pd.concat([top5, pd.Series({'Other': counts.iloc[5:].sum()})])
```

```python
plt.figure(figsize=(7, 7))
plt.pie(pie_data, labels=pie_data.index, autopct='%1.1f%%', startangle=140)
plt.title('Olympic Medal Share by Country (Pie Chart)')
plt.show()
```

```python
plt.figure(figsize=(7, 7))
plt.pie(
    donut_data,
    labels=donut_data.index,
    autopct='%1.1f%%',
    startangle=140,
    wedgeprops={'width': 0.4, 'edgecolor': 'white'}
)
plt.title('Olympic Medal Share by Country (Donut Chart)')
plt.show()
```
